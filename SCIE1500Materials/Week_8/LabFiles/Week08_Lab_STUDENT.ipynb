## 📋 Scientific Scenario: A Conservation Decision

![Lotka-Volterra phase plane and population oscillations](https://github.com/ahailu95/scie1500-content/blob/main/SCIE1500Materials/Week_8/images/W8_predator_prey.svg?raw=1)

Feral cats threaten numbats in the Western Australian wheatbelt. A conservation team must decide whether reducing cats and relocating numbats is likely to create a lasting recovery, or merely change the timing of future population swings.

We begin with a deliberately simple model. It does not predict every real ecological detail; it gives us a mathematical laboratory for linking assumptions, rates of change, graphs, and a management recommendation.

**ODE** stands for **ordinary differential equation**. An ODE describes how a quantity changes over time; here, $dN/dt$ and $dC/dt$ are the instantaneous rates of change of the numbat and cat populations, determined by their current population sizes.

$$\frac{dN}{dt} = 0.4N - 0.002NC$$

$$\frac{dC}{dt} = -0.3C + 0.001NC$$

where $N$ is the numbat population and $C$ is the feral cat population.

- The $+0.4N$ term: numbats grow at 40%/yr in the absence of cats.
- The $-0.002NC$ term: cats reduce numbat growth through predation.
- The $-0.3C$ term: cats decline at 30%/yr without numbats.
- The $+0.001NC$ term: cat population grows when numbats are available as prey.

The interaction term $NC$ is the model's key idea: predation depends on **both** populations. Few cats or few numbats means fewer encounters; large populations of both make predation stronger.

## Part A: Equilibrium and Nullclines (~20 min)

The **nullclines** are curves where one population's rate of change is zero.

✏️ **A.0 — Derive the nullclines algebraically:**

Set $dN/dt = 0$: factor $N(0.4 - 0.002C) = 0$. Since $N \neq 0$ at the coexistence equilibrium, solve $0.4 - 0.002C = 0$ for $C$ — this is the **numbat nullcline**.

Set $dC/dt = 0$: factor $C(-0.3 + 0.001N) = 0$. Since $C \neq 0$ at coexistence, solve $-0.3 + 0.001N = 0$ for $N$ — this is the **cat nullcline**.

```

0.002C = 0.4
C = 0.4/0.002
C = 200
Cat nullcline (dC/dt = 0):    N =
-0.3 + 0.001N = 0
0.001N = 0.3
N = 0.3/0.001
N = 300
```

Run the cell below to check your algebra.

**Terminology:** The coexistence equilibrium is also called an **equilibrium point** or **fixed point**. At this population pair, both rates of change are zero, so the model leaves the pair unchanged if it starts exactly there.

````python
# A.0 — Check your algebra: coexistence equilibrium
C_star = r_N / a   # from 0.4 - 0.002*C = 0
N_star = d   / b   # from -0.3 + 0.001*N = 0
print(f"Numbat nullcline: C* = {C_star:.0f}")
print(f"Cat nullcline:    N* = {N_star:.0f}")
print(f"Coexistence equilibrium: (N*, C*) = ({N_star:.0f}, {C_star:.0f})")
print(f"Trivial equilibrium:     N = 0, C = 0")
````

## Part B: Phase Plane — Turn Algebra into Movement (~20 min)

A phase plane puts the numbat population on the horizontal axis and the cat population on the vertical axis. Each arrow answers one question: **If the system is at this population pair, in which direction will it move next?**

First identify the coexistence equilibrium, $(N^*,C^*)=(300,200)$, where both rates of change are zero. Then start from a nearby population pair, $(N_0,C_0)=(305,195)$, and use the equations to predict its initial movement before running the first plot.

Next, use the complete phase plane to predict the signs of $dN/dt$ and $dC/dt$ in each of the four regions around $(N^*, C^*)=(300,200)$. Then use the arrows to test your prediction.

**Prompt for your group:** Choose one arrow away from the nullclines. Explain, in words and with the signs of both derivatives, why it points in that direction. This is stronger evidence than simply saying that the plot "goes around".

### B.1 — Trace the coexistence equilibrium

````python
# B.1 — Trace movement from a point near coexistence
N0_phase, C0_phase = 305, 195
dt_phase = 0.01
T_phase = 20
steps_phase = int(T_phase / dt_phase)

N_phase = np.zeros(steps_phase + 1)
C_phase = np.zeros(steps_phase + 1)
t_phase = np.zeros(steps_phase + 1)
N_phase[0], C_phase[0] = N0_phase, C0_phase

for i in range(steps_phase):
    dN = r_N * N_phase[i] - a * N_phase[i] * C_phase[i]
    dC = -d * C_phase[i] + b * N_phase[i] * C_phase[i]
    N_phase[i + 1] = N_phase[i] + dN * dt_phase
    C_phase[i + 1] = C_phase[i] + dC * dt_phase
    t_phase[i + 1] = t_phase[i] + dt_phase

print(f"At (N, C) = ({N0_phase}, {C0_phase}):")
print(f"  dN/dt = {r_N * N0_phase - a * N0_phase * C0_phase:.2f}")
print(f"  dC/dt = {-d * C0_phase + b * N0_phase * C0_phase:.2f}")

fig, ax = plt.subplots(figsize=(7, 5))
ax.plot(N_phase, C_phase, color="steelblue", lw=2,
        label="Trajectory from (305, 195)")
ax.plot(N_star, C_star, "ko", ms=9, label="Coexistence equilibrium")
ax.plot(N0_phase, C0_phase, "ro", ms=8, label="Starting point (305, 195)")
ax.set_xlabel("Numbat Population N")
ax.set_ylabel("Feral Cat Population C")
ax.set_title("Trajectory Near the Coexistence Equilibrium")
ax.legend()
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
````

### B.2 — Complete phase plane

The black point at $(N^*, C^*) = (300, 200)$ is called the **coexistence equilibrium**. At exactly this population pair, the model gives $dN/dt = 0$ and $dC/dt = 0$: numbats and feral cats would both remain at the same population size at that instant. It is therefore a fixed reference point of the model, not a claim that real populations will sit there permanently.

In B.1, we deliberately started slightly away from that point, at $(305, 195)$. Because this starting pair is not at equilibrium, both derivatives are non-zero, so the simulated populations move. The curved line in the phase plane and the two time-series lines show what happens to that one off-equilibrium starting point over time.

The equilibrium still matters because it helps us organise and interpret that motion. It marks the crossing of the two nullclines and divides the plane into regions where the signs of $dN/dt$ and $dC/dt$ differ. The complete vector field below generalises B.1: rather than showing only the path from $(305,195)$, it shows the model's predicted direction of movement from many possible numbat-cat population pairs.

As you inspect the arrows, use the equilibrium point as a reference. Ask: if the system starts above, below, left, or right of $(300,200)$, which population is increasing or decreasing, and why?

````python
# B.2 — Complete phase plane with quiver (vector field)
N_grid = np.linspace(0, 600, 20)
C_grid = np.linspace(0, 400, 20)
NN, CC = np.meshgrid(N_grid, C_grid)

dN_dt = r_N * NN - a * NN * CC
dC_dt = -d * CC + b * NN * CC

# Normalise arrows for display
magnitude = np.sqrt(dN_dt**2 + dC_dt**2) + 1e-6
dN_norm = dN_dt / magnitude
dC_norm = dC_dt / magnitude

fig, ax = plt.subplots(figsize=(8, 6))
ax.quiver(NN, CC, dN_norm, dC_norm, magnitude, cmap="RdYlGn", alpha=0.7)

# Nullclines
ax.axvline(N_star, color="steelblue", lw=2, ls="--", label=f"Cat nullcline: N={N_star:.0f}")
ax.axhline(C_star, color="firebrick", lw=2, ls="--", label=f"Numbat nullcline: C={C_star:.0f}")
ax.plot(N_star, C_star, "ko", ms=14, zorder=5, label=f"Coexistence eq. ({N_star:.0f}, {C_star:.0f})")

ax.set_xlabel("Numbat Population N")
ax.set_ylabel("Feral Cat Population C")
ax.set_title("Feral Cat-Numbat Phase Plane")
ax.legend(fontsize=9)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
````

## Part C: Longer Population Trajectories — Watch the Lag Across Cycles (~20 min)

B.1b and Part C both use time-series plots, but they answer different questions. In B.1b, we followed a point very close to the coexistence equilibrium, $(305,195)$, for 20 years. That small trajectory is a focused demonstration of how an off-equilibrium pair begins to move and how the two populations are related over time.

Here we choose a much more imbalanced starting point, $(N_0,C_0)=(400,80)$, and follow it for 30 years: longer than one full cycle in this model. This makes the repeating predator-prey interplay and the lag between the populations easier to see, rather than only demonstrating the first response to a small disturbance.

The phase plane shows the direction of change; this longer time series shows **when** each population changes across repeated rises and falls. Track the sequence rather than reading each line independently: a rise in numbats can support a later rise in cats, which can later drive numbats down.

**Before running the simulation:** predict which population reaches a peak first for $N_0=400$, $C_0=80$. After plotting, check whether the graph supports your prediction and describe the lag and one repeated pattern in one sentence for your presentation.

````python
# C.1 — Simulate using Euler's method
dt   = 0.01   # time step (years)
T    = 30     # total time (years)
steps = int(T / dt)

N0, C0 = 400, 80   # initial populations

N_traj = np.zeros(steps + 1)
C_traj = np.zeros(steps + 1)
t_traj = np.zeros(steps + 1)

N_traj[0], C_traj[0] = N0, C0

for i in range(steps):
    dN = (r_N * N_traj[i] - a * N_traj[i] * C_traj[i]) * dt
    dC = (-d  * C_traj[i] + b * N_traj[i] * C_traj[i]) * dt
    N_traj[i+1] = N_traj[i] + dN
    C_traj[i+1] = C_traj[i] + dC
    t_traj[i+1] = t_traj[i] + dt

print(f"Initial:  N={N0}, C={C0}")
print(f"Year 10:  N={N_traj[int(10/dt)]:.0f}, C={C_traj[int(10/dt)]:.0f}")
print(f"Year 20:  N={N_traj[int(20/dt)]:.0f}, C={C_traj[int(20/dt)]:.0f}")
````

Plot time series.

````python
# C.2 — Plot time series
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6), sharex=True)

ax1.plot(t_traj, N_traj, "steelblue", lw=2, label="Numbats N(t)")
ax1.axhline(N_star, color="steelblue", ls="--", alpha=0.4, label=f"Equilibrium N*={N_star:.0f}")
ax1.set_ylabel("Numbat Population")
ax1.legend(fontsize=9); ax1.grid(alpha=0.3)

ax2.plot(t_traj, C_traj, "firebrick", lw=2, label="Feral Cats C(t)")
ax2.axhline(C_star, color="firebrick", ls="--", alpha=0.4, label=f"Equilibrium C*={C_star:.0f}")
ax2.set_xlabel("Years"); ax2.set_ylabel("Cat Population")
ax2.legend(fontsize=9); ax2.grid(alpha=0.3)

plt.suptitle("Numbat–Feral Cat Population Dynamics (N₀=400, C₀=80)")
plt.tight_layout()
plt.show()
````

## Management Scenario: Intervention Is Not Automatically Improvement

The next simulation adds two ongoing actions: remove 50 cats each year and add 20 numbats each year. This changes the model, so do not assume the original coexistence equilibrium still describes the long-run outcome.

**Before running:** predict what happens if cat numbers reach zero in a model with no carrying capacity for numbats. After plotting, distinguish the **short-term effect** of the intervention from the **long-term behaviour** of this simplified model.

**Presentation prompt:** Make a recommendation with one caveat. State what the simulation suggests, then name one missing ecological mechanism that prevents the result from being a complete real-world forecast.

````python
# C.3 — Management scenario: cat culling + numbat translocation
print("Management scenario: remove 50 cats/year, add 20 numbats/year")
N_mgmt = np.zeros(steps + 1)
C_mgmt = np.zeros(steps + 1)
N_mgmt[0], C_mgmt[0] = 200, 250   # low numbats, high cats (stressed state)

for i in range(steps):
    dN = (r_N * N_mgmt[i] - a * N_mgmt[i] * C_mgmt[i] + 20) * dt
    dC = (-d  * C_mgmt[i] + b * N_mgmt[i] * C_mgmt[i] - 50) * dt
    N_mgmt[i+1] = max(0, N_mgmt[i] + dN)
    C_mgmt[i+1] = max(0, C_mgmt[i] + dC)

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(t_traj, N_mgmt, "steelblue", lw=2, label="Numbats (managed)")
ax.plot(t_traj, C_mgmt, "firebrick", lw=2, label="Feral Cats (managed)")
ax.axhline(N_star, color="steelblue", ls=":", alpha=0.5)
ax.axhline(C_star, color="firebrick", ls=":", alpha=0.5)
ax.set_xlabel("Years"); ax.set_ylabel("Population")
ax.set_title("Managed Scenario: Culling 50 Cats/yr + 20 Numbat Translocations/yr")
ax.legend(); ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
````
