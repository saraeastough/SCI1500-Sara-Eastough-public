### 1(c) — Python: The Vertical Line Test (10 marks)

Consider the four relations below:

1. $y = x^2 - 4x + 3$
2. $x = y^2 - 2$
3. $\dfrac{x^2}{16} + \dfrac{y^2}{9} = 1$
4. $y = \sqrt{16 - x^2}$

**In the code cell below**, plot all four relations (one subplot each, or four separate plots) over a sensible range of $x$ and/or $y$ values. *(6 marks: 1.5 marks per correctly plotted relation — for relations that are not functions of $x$, you will need to plot both branches, e.g. by solving for $y$ in terms of $x$ and plotting $+$ and $-$ roots, or by plotting $x$ as a function of $y$ and swapping axes.)*

**Then, in the Markdown cell below your plot**, state — using the vertical line test — which of the four relations are true functions of $x$ and which are not, with a one-sentence justification for each. *(4 marks)*

<div style="border: 2px solid #2e7d32; border-radius: 8px; padding: 6px 14px; display: inline-block;">✏️ <strong>Write your Python code in the cell below.</strong></div>


```python
import numpy as np
import matplotlib.pyplot as plt

# Relation 1: y = x**2 - 4*x + 3
# Relation 2: x = y**2 - 2   (solve for y: y = +/- sqrt(x + 2))
# Relation 3: x**2/16 + y**2/9 = 1   (an ellipse; solve for y: y = +/- 3*sqrt(1 - x**2/16))
# Relation 4: y = sqrt(16 - x**2)

x_reg = np.linspace(-5, 5, 400)
x_ellipse = np.linspace(-4, 4, 400)
y_inv_para = np.linspace(-3, 3, 400)

fig, axs = plt.subplots(2, 2, figsize=(12, 10))

# Relation 1: y = x**2 - 4*x + 3
y1 = x_reg**2 - 4*x_reg + 3
axs[0, 0].plot(x_reg, y1, 'b-', label=r'$y = x^2 - 4x + 3$')
axs[0, 0].set_title('Relation 1')
axs[0, 0].set_xlim([-2, 6])
axs[0, 0].set_ylim([-2, 8])

# Relation 2: x = y**2 - 2  ->  y = +/- sqrt(x + 2)
x2 = y_inv_para**2 - 2
axs[0, 1].plot(x2, y_inv_para, 'r-', label=r'$x = y^2 - 2$')
axs[0, 1].set_title('Relation 2')
axs[0, 1].set_xlim([-3, 5])
axs[0, 1].set_ylim([-4, 4])


# Relation 3: x**2/16 + y**2/9 = 1 ->  y = +/- 3*sqrt(1 - x**2/16)
y3_top = 3 * np.sqrt(1 - (x_ellipse**2 / 16))
y3_bottom = -3 * np.sqrt(1 - (x_ellipse**2 / 16))
axs[1, 0].plot(x_ellipse, y3_top, 'g-', label='Top branch')
axs[1, 0].plot(x_ellipse, y3_bottom, 'g--', label='Bottom branch')
axs[1, 0].set_title('Relation 3')
axs[1, 0].set_xlim([-5, 5])
axs[1, 0].set_ylim([-4, 4])
axs[1, 0].set_aspect('equal')


# Relation 4: y = sqrt(16 - x**2)
y4 = np.sqrt(16 - x_ellipse**2)
axs[1, 1].plot(x_ellipse, y4, 'm-', label=r'$y = \sqrt{16 - x^2}$')
axs[1, 1].set_title('Relation 4')
axs[1, 1].set_xlim([-5, 5])
axs[1, 1].set_ylim([-1, 5])
axs[1, 1].set_aspect('equal')

for ax in axs.flat:
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)
    ax.set_xlabel('$x$')
    ax.set_ylabel('$y$')
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

<div style="border: 2px solid #2e7d32; border-radius: 8px; padding: 10px 16px;">

**Your classification and justification for each relation:**

**Relation 1** ($y = x^2 - 4x + 3$):
It is a true function of x. Because it passes through the vertical line only once.

**Relation 2** ($x = y^2 - 2$):
This is not a true function of x. Becuase there are two points that cross through the vertical line.

**Relation 3** ($\dfrac{x^2}{16} + \dfrac{y^2}{9} = 1$):
This is not a true function of x. Because there are two points where the line crosses the verticle line through the ellipse.

**Relation 4** ($y = \sqrt{16 - x^2}$):
True function of x. Because there is only 1 intersection through the vertical line.

### 4(c) — Python: Visualising the dosing profile (9 marks)

**In the code cell below**, define $C(t)$, then plot the concentration profile for $t \in [0, 24]$ hours. Mark the peak point found in 4(a), and shade the therapeutic window (5–40 $\mu g/mL$). *(6 marks)*

**Then, in the Markdown cell below your plot**, state approximately how many hours the concentration stays within the therapeutic window, and whether a single dose is adequate to cover a typical 8-hour period between doses. *(3 marks)*

<div style="border: 2px solid #2e7d32; border-radius: 8px; padding: 6px 14px; display: inline-block;">✏️ <strong>Write your Python code in the cell below.</strong></div>


```python
import numpy as np
import matplotlib.pyplot as plt

D, k1, k2 = 60, 0.15, 0.6

def C(t):
    return D * (np.exp(-k1 * t) - np.exp(-k2 * t))

t_vals = np.linspace(0, 24, 500)
c_vals = C(t_vals)

# t_peak = ln(k2/k1) / (k2 - k1)
t_peak = np.log(k2 / k1) / (k2 - k1)
c_peak = C(t_peak)

plt.figure(figsize=(10, 6))
plt.plot(t_vals, c_vals, label="Concentration $C(t)$", color="blue", lw=2)

plt.scatter(t_peak, c_peak, color="red", zorder=5, label=fr"Peak Point ({t_peak:.2f}h, {c_peak:.2f} $\mu$g/mL)")

plt.axhspan(5, 40, color="green", alpha=0.15, label=r"Therapeutic Window (5-40 $\mu$g/mL)")

plt.title("Dosing Profile & Concentration Over Time")
plt.xlabel("Time (hours)")
plt.ylabel(r"Concentration ($\mu$g/mL)")
plt.xlim(0, 24)
plt.ylim(0, 50)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()

plt.show()
```
