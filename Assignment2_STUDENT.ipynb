**In the code cell below**, use `scipy.integrate.quad` to numerically compute the **total** canopy area added by Site A and by Site B over the first $t = 0$ to $t = 8$ years (i.e. $\int_0^8 A'(t)\,dt$ and $\int_0^8 B'(t)\,dt$), and print both totals together with a statement of which site added more canopy overall.


<div style="border: 2px solid #2e7d32; border-radius: 8px; padding: 6px 14px; display: inline-block;">✏️ <strong>Write your Python code in the cell below.</strong></div>

```python
from scipy import integrate
import numpy as np

def Aprime(t):
    return 6 - 0.4 * t + 5 / (t + 2)

def Bprime(t):
    return 8 * np.exp(-0.15 * t)

# Compute the definite integrals from t = 0 to t = 8
total_added_A, _ = integrate.quad(Aprime, 0, 8)
total_added_B, _ = integrate.quad(Bprime, 0, 8)

# Print both totals
print(f"Total canopy area added by Site A: {total_added_A:.3f} thousand m^2")
print(f"Total canopy area added by Site B: {total_added_B:.3f} thousand m^2")

# Comparison statement
if total_added_A > total_added_B:
    print("Site A added more canopy overall.")
elif total_added_B > total_added_A:
    print("Site B added more canopy overall.")
else:
    print("Both sites added the exact same amount of canopy.")
```

with parameters $\alpha = 0.5$, $\beta = 0.02$, $\lambda = 0.005$, $\gamma = 0.4$ (all per year, except $\beta$ and $\lambda$ which are per hundred individuals per year).

### 4(a) — Finding the equilibrium points (6 marks)

Set $\dfrac{dH}{dt} = 0$ and $\dfrac{dP}{dt} = 0$ and solve algebraically to find **both** equilibrium points $(H,P)$ — the trivial one and the non-trivial (biologically interesting) one. ### 4(c) — Python: Simulating the system (10 marks)

Starting from $H(0) = 50$ and $P(0) = 20$ (hundreds of individuals):

(i) **In the code cell below**, use `scipy.integrate.odeint` (or `solve_ivp`) to simulate the system for $t \in [0, 100]$ years. Plot $H(t)$ and $P(t)$ **together on one time-series graph**, with axis labels and a legend. *(5 marks)*

(ii) Produce a second plot: a **phase portrait** (hare population $H$ on the $x$-axis, lynx population $P$ on the $y$-axis) showing the trajectory traced out by the system, and mark the non-trivial equilibrium point found in 4(a) on this plot. *(3 marks)*

(iii) In the Markdown cell below your plots, describe in one or two sentences whether the system appears to **cycle** around the equilibrium or **converge** to it. *(2 marks)*


<div style="border: 2px solid #2e7d32; border-radius: 8px; padding: 6px 14px; display: inline-block;">✏️ <strong>Write your Python code in the cell below.</strong></div>

```python
import numpy as np
from scipy.integrate import odeint
import matplotlib.pyplot as plt

alpha, beta, lam, gamma = 0.5, 0.02, 0.005, 0.4

def lotka_volterra(state, t):
    H, P = state
    dHdt = alpha * H - beta * H * P
    dPdt = lam * H * P - gamma * P
    return [dHdt, dPdt]

t = np.linspace(0, 100, 2000)
state0 = [50, 20]

# Parameters
alpha, beta, lam, gamma = 0.5, 0.02, 0.005, 0.4

# Define the Lotka-Volterra system of differential equations
def lotka_volterra(state, t):
    H, P = state
    dHdt = alpha * H - beta * H * P
    dPdt = lam * H * P - gamma * P
    return [dHdt, dPdt]

# Time points (t in [0, 100] years)
t = np.linspace(0, 100, 2000)
state0 = [50, 20]  # Initial conditions H(0)=50, P(0)=20

# 1. Solve the system with odeint
sol = odeint(lotka_volterra, state0, t)
H = sol[:, 0]
P = sol[:, 1]

# 2. Plot H(t) and P(t) together on one time-series graph
plt.figure(figsize=(10, 4))
plt.plot(t, H, label='Snowshoe Hare ($H$)', color='forestgreen')
plt.plot(t, P, label='Canadian Lynx ($P$)', color='firebrick')
plt.xlabel('Time ($t$ in years)')
plt.ylabel('Population (hundreds)')
plt.title('Lotka-Volterra: Population Dynamics Over Time')
plt.legend()
plt.grid(True)
plt.show()

# 3. Produce a second plot: phase-portrait (H vs P)
plt.figure(figsize=(6, 6))
plt.plot(H, P, label='Trajectory', color='navy')
plt.plot(80, 25, 'ro', label='Non-trivial Equilibrium (80, 25)')
plt.xlabel('Snowshoe Hare Population ($H$)')
plt.ylabel('Canadian Lynx Population ($P$)')
plt.title('Phase Portrait: Hare vs. Lynx')
plt.legend()
plt.grid(True)
plt.show()
```
